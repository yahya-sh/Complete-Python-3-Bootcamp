# Guessing Game Challenge

The Challenge:

Write a program that picks a random integer from 1 to 100, and has players guess the number. The rules are:

1. If a player's guess is less than 1 or greater than 100, say "OUT OF BOUNDS"
2. On a player's first turn, if their guess is
 * within 10 of the number, return "WARM!"
 * further than 10 away from the number, return "COLD!"
3. On all subsequent turns, if a guess is 
 * closer to the number than the previous guess return "WARMER!"
 * farther from the number than the previous guess, return "COLDER!"
4. When the player's guess equals the number, tell them they've guessed correctly *and* how many guesses it took!

Note: `random.randint(a,b)` returns a random integer in range `[a, b]`, including both end points.

#### Write a `while` loop that asks for a valid guess. Test it a few times to make sure it works.

```python
guess_count = 0
last_guess = None
while True:
    guess = int(input("Guess a number:"))
    guess_count+=1
    if guess < bound_start or guess > bound_end:
        print("OUT OF BOUNDS")
        continue
    if guess == x:
        print("You've guessed the number correctly!")
        print("You guessed the number after {} tries!".format(guess_count))
        break
    first_run = last_guess == None
    difference = abs(guess-x)
    if first_run:
        warm = difference <= 10
        if warm:
            print("WARM!")
        else:
            print("COLD!")
    else:
        last_guess_difference = abs(last_guess-x)
        if difference < last_guess_difference:
            print("WARMER!")
        else:
            print("COLDER!")

    last_guess = guess
            
    
    
        
```
