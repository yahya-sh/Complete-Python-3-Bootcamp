```python
def count(start=0, step=1):
    while True:
        yield start
        start+=step


start = 30
for i in count(start, 1):
    print(i)
    if i >= start + 10:
        break
```

```python
from itertools import chain
nested = [ 
    [0, 1],
    [2, 3, 4],
    [5],
]

print(list(chain(*nested)))
print(list(chain.from_iterable(nested)))
```
